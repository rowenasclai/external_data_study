```python
import requests
from bs4 import BeautifulSoup

url = "https://www.ccmtshow.com/311/list.html"
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')

# Adjust selectors based on actual page structure
exhibitors = soup.find_all('tr')  # or 'div', 'li', etc.
for row in exhibitors:
    name = row.find('td', class_='el-table_1_column_2  ')  # adjust selector
    booth = row.find('td', class_='el-table_1_column_3  ')   # adjust selector
    if name and booth:
        print(f"Exhibitor: {name.text}, Booth: {booth.text}")
```
