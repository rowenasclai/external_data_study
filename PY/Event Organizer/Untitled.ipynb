```python
from pathlib import Path

folder_path = Path("/Users/rowena/Other Projects/external_data_study/Result/Exhibition Organizers/cantonfair/2026")

df_list = []
all_records = []
tmp=pd.DataFrame()
for file_path in folder_path.glob("*.json"):
    # read_json handles arrays of objects or record-oriented JSON
    #df = pd.read_json(file_path,lines=True)

    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            all_records.extend(json.loads(line))

# Build single combined DataFrame
    df = pd.DataFrame(all_records)
    
    # Optional: track the source filename
    df["source_file"] = file_path.name

    tmp=pd.concat([tmp, df],axis=0)
    
    #df_list.append(df)
```
