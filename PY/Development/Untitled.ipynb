```python
import pandas as pd
from glob import glob
import os

# Define the path to the directory containing the JSON files
# Use an absolute or relative path, e.g., 'path/to/json_folder/'
path_to_json_files = '/Users/rowena/Other Projects/external_data_study/Result/cantonfair/' 

# Find all files ending with '.json' in the specified directory
json_files = glob(os.path.join(path_to_json_files, '*.json'))

# Read each file into a DataFrame and concatenate them into a single DataFrame
# The list comprehension efficiently processes all files
df = pd.concat((pd.read_json(f, orient='records', lines=True) for f in json_files), ignore_index=True)

# Display the combined DataFrame
#print(df.head())
```

```python
df.to_csv(path_to_json_files+'cantonfair.csv',encoding='utf-8-sig',index=False)
```
